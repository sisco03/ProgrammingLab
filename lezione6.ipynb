```python
import pandas as pd

sales = pd.DataFrame(
    data={
        "employee": [
            "Katrina",
            "Guanyu",
            "Jan",
            "Roman",
            "Jacqueline",
            "Paola",
            "Esperanza",
            "Alaina",
            "Egweyn",
        ],
        "sales": [14, 17, 6, 12, 8, 3, 7, 15, 5],
        "year": [2018, 2019, 2020, 2018, 2020, 2019, 2019, 2020, 2020],
    }
)
sales

#4.Rispondere alle domande

#Mostrami le vendite maggiori di 10

vendite_maggiori_10 = sales.query('sales > 10')

#Mostrami i dati del 2018

dati_2018 = sales.query('year == 2018')

#Mostrami le vendite maggiori di 13 e l'anno è 2018

maggiori_13_anno_2018 = sales.query('sales > 13 & year == 2018')

#Mostrami TUTTO tranne i casi in cui le vendite sono maggiori di 13 e l'anno è 2018

tutto_tranne_maggiori_13_anno_2018 = sales.query('sales < 13 & year != 2018')

#Mostrami i dati dove le vendite divise per 3 sono maggiori di 3

vendite_diviso_3_maggiori_3 = sales.query('(sales/3) > 3')

#Mostrami i dipendenti i cui nomi sono alfabeticamente dopo la J

sales.set_index('employee', inplace=True)
#riordino alfabeticamente
sales.sort_index(inplace=True)

dipendenti_dopo_j = sales.query('employee > "K"')

#risultati
vendite_maggiori_10, dati_2018, maggiori_13_anno_2018, tutto_tranne_maggiori_13_anno_2018, vendite_diviso_3_maggiori_3, dipendenti_dopo_j
```

```python
#5.Analisi di un dataset di video giochi
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)


#1.Quanti videogiochi sono stati pubblicati

#ogni riga rappresenta un gioco diverso
tot_giochi = df.shape[0]

#2.Quali sono i generi più popolari

generi_popolari = df['Genre'].value_counts()

#3.Plotta l'evoluzione del numero di giochi nel tempo

plt.hist(df['Year'], bins=20, linewidth=0.5, alpha=0.5, edgecolor='black')
plt.title('Evoluzione numero di giochi nel tempo')
plt.xlabel('anno')
plt.ylabel('totale giochi')
plt.show()

#4.Prova a riprodurre questo plot

top_5_generi = df['Genre'].value_counts().head(5)

colonne = ['EU_Sales', 'JP_Sales', 'NA_Sales', 'Other_Sales']

colori = ['blue', 'orange', 'green', 'red']

#filtro
subset = df[df['Genre'].isin(top_5_generi.index)]
#aggrego le vendite sommando per genere (solo i top 5)
vendite = subset.groupby('Genre')[colonne].sum()
#per tenere traccia dove finisce il segmento ed inizia il prossimo
bottom = np.zeros(len(vendite))

for colonna, colore in zip(colonne, colori):
    
    plt.bar(vendite.index, vendite[colonna], bottom=bottom, color=colore, label=colonna)
    #aggiorno bottom
    bottom += vendite[colonna].values


plt.title('Vendite per regione per genere')
plt.xlabel('Genre')
plt.ylabel('Milioni di unità')
plt.legend()
plt.show()

#risultati
tot_giochi, generi_popolari



```
